```python
import numpy as np

arr = np.arange(1, 11)

print("Array:", arr)
print("Type:", type(arr))
print("Shape:", arr.shape)
```

```python
zeros = np.zeros((3, 3))
ones = np.ones((3, 3))

print("Zeros Matrix:")
print(zeros)

print("Ones Matrix:")
print(ones)
```

```python
arr = np.linspace(0, 1, 10)

print(arr)
```

```python
arr = np.random.randint(1, 101, 10)
print("Original Array:", arr)
sorted_arr = np.sort(arr)
print("Sorted Array:", sorted_arr)
```

```python
arr = np.array([12, 45, 23, 56, 78, 10])
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))
print("Mean:", np.mean(arr))
print("Standard Deviation:", np.std(arr))
```

```python
arr = np.arange(1, 13)
matrix = arr.reshape(3, 4)
print(matrix)
```

```python
arr = np.array([1, 2, 3, 4, 5])
print("Original:", arr)
print("Add 5:", arr + 5)
print("Multiply by 2:", arr * 2)
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Dot Product:", np.dot(a, b))
```

```python
arr = np.array([10, 20, 30, 40, 50, 60, 70, 80])
print("First 3:", arr[:3])
print("Last 3:", arr[-3:])
print("Middle Elements:", arr[3:5])
```
