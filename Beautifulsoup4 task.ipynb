```python
#pip install requests beautifulsoup4
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://quotes.toscrape.com/"
```

```python
content = requests.get(url)
content
```

```python
bs = BeautifulSoup(content.text, "html.parser")
q = bs.find_all("span", class_="text")
q
```

```python
for i in q:
    print(i.text)
```
