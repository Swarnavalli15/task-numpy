```python
import pandas as pd
```

```python
df = pd.read_csv("Bank-Customers.csv")
print(df.head())
```

```python
X = df[["Earning", "Credit Score"]]
print(X.head())
```

```python
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt
```

```python
S = sch.dendrogram(sch.linkage(X, method="ward"))
plt.show()
```

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
cm = AgglomerativeClustering(n_clusters=4,linkage="ward")
y_pred = cm.fit_predict(X)
```

```python
plt.scatter( X.iloc[:, 0], X.iloc[:, 1],c=y_pred,cmap="rainbow")
plt.show()

```
